# 01 Table VI Exploration

Initial exploration of the manually curated Table VI dataset.


```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/curated/e1020_table_vi_v0.csv')
df.head()
```

```python
df['iv_potency_class'].value_counts()
```

```python
ranking = df[['compound_id', 'ed50_ug_per_kg', 'ed50_relation', 'iv_potency_class']].sort_values('ed50_ug_per_kg')
ranking.head(15)
```

```python
plot_df = df[df['ed50_relation'] == '='].copy().sort_values('ed50_ug_per_kg')
ax = plot_df.plot.bar(x='compound_id', y='ed50_ug_per_kg', legend=False, figsize=(12, 4))
ax.set_ylabel('ED50 (μg/kg)')
ax.set_title('Exact ED50 values from Table VI')
plt.tight_layout()
```
